```python
import multiprocessing
import time
import random
from datetime import datetime


def worker():
    lunch = random.uniform(0, 10)

    print(
        f"{multiprocessing.current_process().name} "
        f"is taking a {lunch:.2f} second break",
        flush=True
    )

    time.sleep(lunch)

    break_over = datetime.now()

    print(
        f"{multiprocessing.current_process().name} "
        f"was back from break at {break_over.strftime('%H:%M:%S')}",
        flush=True
    )


# Jupyter is a pain
ctx = multiprocessing.get_context("fork")

break_times = []

for i in range(3):
    labor = ctx.Process(
        target=worker,
        name=f"Laborer-{i + 1}"
    )

    break_times.append(labor)
    labor.start()

for labor in break_times:
    labor.join()

print("All laborers are back from break.")
```
