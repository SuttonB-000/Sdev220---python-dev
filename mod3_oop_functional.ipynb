```python
class Solution:
    def sort012(self, arr):
        #implementing a selction sort
        n = len(arr)
        for i in range(n-1):
            min_idx = i
            for j in range(i + 1, n):
                if arr[j] < arr[min_idx]:
                    min_idx = j
            # swapping 
            arr[i], arr[min_idx] = arr[min_idx], arr[i]
        return
```

```python
class Solution:
    def firstSearch(self, arr, k):
        # Code Here
        length = len(arr)
        for i in range(length):
            if arr[i] == k:
                return i
        else:
            return -1
            
```
